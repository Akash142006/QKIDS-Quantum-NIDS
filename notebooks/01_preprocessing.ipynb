## Configuration

**IMPORTANT**: Update the paths below to point to your NSL-KDD dataset files!

```python
# ============================================
# CONFIGURATION - UPDATE THESE PATHS!
# ============================================

TRAIN_PATH =  'D:/QKIDS/QKIDS/data/KDDTrain+.txt' # Update this path
TEST_PATH = 'D:/QKIDS/QKIDS/data/KDDTest+.txt'
   # Update this path

SAMPLE_SIZE = 1000  # Total samples (800 train + 200 test)
QUBIT_COUNTS = [4, 5, 6, 7, 8]  # Qubit counts to test
```

## Preprocess Data for Each Qubit Count

```python
# Initialize preprocessor
preprocessor = NSLKDDPreprocessor(
    train_path=TRAIN_PATH,
    test_path=TEST_PATH,
    sample_size=SAMPLE_SIZE,
    random_state=42
)

# Store all preprocessed datasets
preprocessed_data = {}

# Preprocess for each qubit count
for n_qubits in QUBIT_COUNTS:
    print_section(f"Processing {n_qubits} Qubits")
    
    # Preprocess
    data = preprocessor.preprocess_for_qubit_count(n_qubits)
    
    # Save
    filename = f'preprocessed_{n_qubits}qubits.pkl'
    preprocessor.save_preprocessed_data(data, filename)
    
    # Store in dictionary
    preprocessed_data[n_qubits] = data
    
    print(f"\n✅ {n_qubits} qubits preprocessing complete!\n")
```

## Summary of Preprocessed Data

```python
import pandas as pd

print("\n" + "="*70)
print("  PREPROCESSING SUMMARY")
print("="*70 + "\n")

summary_data = []
for n_qubits, data in preprocessed_data.items():
    summary_data.append({
        'Qubits': n_qubits,
        'Train Samples': len(data['X_train']),
        'Test Samples': len(data['X_test']),
        'Features': len(data['selected_features']),
        'Top Feature': data['selected_features'][0]
    })

summary_df = pd.DataFrame(summary_data)
print(summary_df.to_string(index=False))

print("\n✅ All preprocessing complete! Ready for training.")
print("\n📁 Preprocessed files saved in: ../data/")
print("\n➡️  Next step: Run notebook 02_train_evaluate.ipynb")
```
